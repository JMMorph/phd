```python
import numpy as np
import pandas as pd
import geopandas as gpd

from sklearn.preprocessing import RobustScaler

files_path = 'files/'
```

```python
metadata = pd.read_csv(files_path + 'timeseries_metadata.csv')
data = pd.read_csv(files_path + 'timeseries_data.csv', parse_dates=['datetime'])
```

# Visualize Map

```python
# we assume that each station has only one lat/lon coordinate
stations = metadata.groupby('station_name').agg({'lat': 'first', 
                                      'lon': 'first',
                                      'variable_name': set}).reset_index()

gdf = gpd.GeoDataFrame(stations, 
                       geometry=gpd.points_from_xy(stations.lon, stations.lat),
                       crs="EPSG:4326")
```

```python
# convert unhashable sets to a hashable representation (frozenset) before counting/mapping
var_key = gdf['variable_name'].apply(lambda s: frozenset(s) if isinstance(s, set) else s)
# map every unique combination of variables to a unique index
var_to_index = {var: str(idx) for idx, var in enumerate(var_key.unique())}

gdf['variable_index'] = var_key.map(var_to_index)
gdf['variables'] = var_key.apply(lambda s: ', '.join(list(s)) if isinstance(s, frozenset) else s)
```

```python
import plotly.graph_objects as go
import plotly.express as px

# Create a scatter map using plotly
fig = px.scatter_map(
    gdf,
    lat='lat',
    lon='lon',
    text=gdf['station_name'].apply(lambda x: '<br><br>'+x),
    hover_name='station_name',
    hover_data={'lat': False, 'lon': False, 'variable_index': False},
    color='variables',
    zoom=10.5,
    title='Air Quality Monitoring Stations',
    map_style="carto-positron-nolabels",
    size = gdf['variable_name'].apply(lambda x: len(x)-1)
)




# Update layout
fig.update_layout(
    width=1200,
    height=1000
)

fig.show()
```

# Visualize Heatmap + gaps

```python
data.set_index('datetime', inplace=True)
# Resample hourly by station and variable, filling gaps with NaN
data = (
    data
    .groupby(["variable", "station"])
    .resample("h")
    .asfreq()['value']   
    .reset_index()
)

```

```python
data = data.pivot_table(index=['variable', 'datetime'], columns='station', values='value', aggfunc='first', dropna=False).reset_index()
```

```python
# Create a copy of the dataframe
df_consecutive_nans = data.copy()

# Group by the 'variable' column and sort by 'date'
df_consecutive_nans.sort_values(by='datetime', inplace=True)
grouped = df_consecutive_nans.groupby('variable')

# Iterate over each group to calculate consecutive NaNs
for name, group in grouped:
    for col in group.columns:
        if col not in ['variable', 'datetime']:
            # Identify NaN values
            is_nan = group[col].isna()
            
            # Group consecutive NaNs and calculate their count
            nan_group = (is_nan != is_nan.shift()).cumsum()
            df_consecutive_nans.loc[group.index, col] = is_nan.groupby(nan_group).transform('sum')
            
            # Set non-NaN values to 0
            df_consecutive_nans.loc[group.index[~is_nan], col] = 0

# Convert all columns except 'year' and 'month' to integers
columns_to_convert = [col for col in df_consecutive_nans.columns if col not in ['variable', 'datetime',]]
df_consecutive_nans[columns_to_convert] = df_consecutive_nans[columns_to_convert].astype(int)

df_temp_gaps = df_consecutive_nans.copy()

# Define the bins and corresponding labels
bins = [0, 1, 3, 12, 24, 24*3, 24*7, 24*30, 24*30*3, 24*30*6, np.inf]
labels = np.arange(len(bins) - 1)

for column in df_temp_gaps.columns:
    if column not in ['variable', 'datetime']:
        # Replace 0 with NaN for the purpose of binning
        df_temp_gaps[column] = np.digitize(df_temp_gaps[column], bins, right=True) 
```

```python
variable = 'o3'
year = 2017

minidf = data.query(f"(variable == '{variable}')&(datetime.dt.year=={year})").drop(columns = ['variable'])
minidf_gaps = df_temp_gaps.query(f"(variable == '{variable}')&(datetime.dt.year=={year})").drop(columns = ['variable'])
minidf_gaps = minidf_gaps.replace({0: np.nan})
minidf_gaps.set_index('datetime', inplace=True)
minidf_gaps.sort_index(inplace=True)


for col in minidf.columns:
    # replace string 'nr' with np.nan
    minidf[col] = minidf[col].replace({'nr': np.nan})
    if col != 'datetime':
        minidf[col] = pd.to_numeric(minidf[col])
    if minidf[col].isna().all():
        minidf.drop(columns=[col], inplace=True)
        

minidf.set_index('datetime', inplace=True)
minidf.sort_index(inplace=True)

total_nans_by_column = minidf.isna().sum()
columns_by_nans = total_nans_by_column.sort_values(ascending=True).index

minidf = minidf[columns_by_nans]
minidf_array = minidf.copy()
array = minidf_array.values

minidf_gaps = minidf_gaps[columns_by_nans]
minidf_array_gaps = minidf_gaps.copy()
array_gaps = minidf_array_gaps.values

# scaler = RobustScaler()
# minidf_color = minidf.copy()
# minidf_color = scaler.fit_transform(minidf_color)

minidf_color = minidf.transform(lambda x: (x - x.min()) / (x.max() - x.min()))

array_color = minidf_color.values

estaciones = minidf.columns.tolist()
```

```python
colormap_values = 'GnBu_r'
colormap_gaps = 'Burgyl'

fig_width = 1600
fig_height = len(minidf_array.columns)*30 + 300
```

```python
# -----------------------------------------------------------------------------------------
# General figure setup


# Create empty figure with y inverted for better visualization
fig = go.Figure()
fig.update_yaxes(autorange="reversed")

family_font = "Times New Roman"


# # --- Timestamps each X hours
# # Add x axis labels just for every 24 hours (change as needed)
# hours = 24
# tickvals=minidf_array.index[::hours]
# ticktext=minidf_array.index[::hours].strftime('%Y-%m-%d')

# --- Timmestamps each X day
days = ['1', '15']
tickindex=minidf_array.index.day.astype(str).isin(['1', '15'])\
    & (minidf_array.index.hour == 0)
tickvals=minidf_array.index[tickindex]
ticktext=minidf_array.index[tickindex].strftime('%Y-%m-%d')

fig.update_xaxes(tickvals= tickvals, ticktext= ticktext)


# Small window of 31 days (change as needed)
start = minidf_array.index.min().strftime(f'{year}-01-01')
end = minidf_array.index.min() + pd.Timedelta(days=31)
fig.update_layout(xaxis_range=[start, end])

# Adjust scale and labels
fig.update_layout(width=fig_width, 
                  height=fig_height)

# Update axis labels
fig.update_xaxes(title_text='← Time →',
                 title_font=dict(size=20, family=family_font),
                 tickfont=dict(size=16, family=family_font),
                 tickangle=30
                 )
fig.update_yaxes(title_text='← Stations →',
                 title_font=dict(size=20, family=family_font),
                 tickfont=dict(size=16, family=family_font)
                 )

# Ad month annotations 
month_start = pd.date_range(start=start, periods=12, freq="MS")  

# Add a vertical line end of month
for inicio, middle in zip(month_start, 
                          month_start + pd.Timedelta(days=14)):
    fig.add_vline(x=inicio, 
                  line_width=2, 
                  line_dash="dashdot", 
                  line_color="red")
    

    fig.add_annotation(
    x=middle,
    yref="paper", y=1.05,  # posición relativa al canvas
    text=middle.strftime('%B').capitalize(),
    showarrow=False,
    font=dict(size=20, color="gray", family=family_font),
    xanchor="center"
    )


# Change background to transparent (both paper and plot, change as needed)
fig.update_layout(paper_bgcolor='rgba(0,0,0,0)', 
                  plot_bgcolor='rgba(0,0,0,0)')





# -----------------------------------------------------------------------------------------
# Plot values

values_fig = px.imshow(array_color.T, 
                aspect='auto', 
                title=f'Visualización {variable.upper()}',
                x=minidf_array.index,
                y=minidf_array.columns,
                )

# Add hovertemplate to include array_color information
values_fig.update_traces(hovertemplate='''
                         Date: %{x}<br>
                         Station: %{y}<br>
                         Value: %{customdata}''')

# Add customdata to the figure
values_fig.data[0].update(customdata=np.where(np.isnan(array.T), 
                                              'Gap ' + array_gaps.T.astype(int).astype(str), 
                                              array.T.astype(str)))

# First trace use color axis (coloraxis1)
values_fig.update_traces(
    showscale=True,
    coloraxis='coloraxis1'
)





# -----------------------------------------------------------------------------------------
# Plot gaps

gaps_fig = px.imshow(
    array_gaps.T,
    x=minidf_array_gaps.index,
    y=minidf_array_gaps.columns,
)

# Add hovertemplate to include array_color information
gaps_fig.update_traces(hovertemplate='''
                         Date: %{x}<br>
                         Station: %{y}<br>
                         Value: Gap %{z}''')


# Second trace use different color axis (coloraxis2)
gaps_fig.update_traces(
    showscale=True,
    coloraxis='coloraxis2'
)





# -----------------------------------------------------------------------------------------
# Add colorbars

colorbars_font = titlefont=dict(size=20, 
                                color="gray",
                                family="Consolas")
colorbars_tickfont = tickfont=dict(size=17,
                                  color="gray",
                                  family="Consolas")

# Colorbar for values
fig.update_layout(
    coloraxis1=dict(
        colorbar=dict(title="Values",
                      x=0.5, orientation='h', y=-0.3, yref = "paper",
                      tickvals=np.linspace(0, 1, 11),
                      ticktext=np.linspace(minidf.min().min(), 
                                           minidf.max().max(), 
                                           11).round(2).astype(str).tolist(),
                      titlefont=colorbars_font,
                      tickfont=colorbars_tickfont
                      ),
        colorscale=colormap_values # values
        
    ))

# Colorbar for gaps
fig.update_layout(
    coloraxis2=dict(
        colorbar=dict(title='_Gaps_', 
                      x=0.5, orientation='h', y=-0.4, yref = "paper",
                      titlefont=colorbars_font,
                      tickfont=colorbars_tickfont
                      ),
        colorscale=colormap_gaps,  # gaps
    ),
)

# Add plots to the main figure
for trace in gaps_fig.data:
    fig.add_trace(trace)
for trace in values_fig.data:
    fig.add_trace(trace)



# Add visibility menu
fig.update_layout(
    updatemenus=[
        dict(
            type="dropdown",
            direction="down",
            x=-0.1,
            y=1,
            buttons=[
                dict(label="Values & Gaps", method="restyle", args=[{"visible": [True, True]}]),
                dict(label="Values only", method="restyle", args=[{"visible": [False, True]}]),
                dict(label="Gaps only", method="restyle", args=[{"visible": [True, False]}])
            ]
        )
    ]
)


fig.show()
fig.write_html(f"output/heatmap_{variable}_{year}.html")
```
